```python
import cv2
import os, sys
from ultralytics import YOLO
from pathlib import Path
```

```python
# Prevents the OpenMP multithreading loop from triggering access violation status_access_violation
os.environ["KMP_DUPLICATE_LIB_OK"] = "TRUE"
```

```python
# Check if a previous training run exists and has a checkpoint saved
checkpoint_path = Path("c:/Users/user/OneDrive/Desktop/Cilent Side/weapon-Detection-YOLOV26/.ipynb_checkpoints/runs/detect/weapon_detection/pistol_only_run/weights/last.pt")

if checkpoint_path.exists():
    print(f" Found existing training checkpoint at: {checkpoint_path}")
    print("RESUMING TRAINING FROM WHERE IT PAUSED...")
    
    # Load the partially trained weights directly
    model = YOLO(str(checkpoint_path))
    
    # Resume training. Note: Parameters like data, imgsz, and batch are 
    # locked automatically from your original configuration to prevent corruption.
    results = model.train(resume=True)

else:
    print("No previous checkpoint found. STARTING FRESH NEW TRAINING...")

    model = YOLO("yolo26n.pt")
    results = model.train(
        data="../dataset.yaml",
        epochs=50,
        imgsz=640,
        batch=8,
        cache=False,
        workers=2,
        device='cpu',
        project="weapon_detection",
        name="pistol_only_run"
    )
```

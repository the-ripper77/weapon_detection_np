```python

# 1. Load your champion model
model = YOLO(model_path)

# Extract configurations once to save loop overhead
class_name = model.names[0]  # Hardcoded since it's a 1-class 'pistol' model
box_color = (0, 0, 255)      # BGR Red

# Pre-initialize tracking variables so they exist on frame 1
boxes = []
confs = []

# 2. Optimize Webcam Capture Settings directly via hardware hooks
cap = cv2.VideoCapture(0, cv2.CAP_DSHOW) # CAP_DSHOW cuts camera startup lag on Windows
cap.set(cv2.CAP_PROP_FRAME_WIDTH, 640)   # Ask webcam for standard matrix size
cap.set(cv2.CAP_PROP_FRAME_HEIGHT, 480)

window_name = "YOLO26 Optimized Detection"
cv2.namedWindow(window_name, cv2.WINDOW_NORMAL)

print("🚀 High-Performance OpenCV Stream Starting...")
print("👉 Click on the window and press 'q' to stop safely.")

frame_count = 0
fps = 0
prev_time = time.time()

try:
    while cap.isOpened():
        ret, frame = cap.read()
        if not ret:
            break
        
        frame_count += 1
        
        # Frame Skipping: Run AI inference only on even frames to double smoothness
        if frame_count % 2 == 0:
            # Set imgsz=480 for the perfect speed-to-distance balance
            results = model(frame, imgsz=480, conf=0.40, device='cpu', verbose=False)
            prediction = results[0]
            
            # FIX: Extract raw tensor data arrays correctly and efficiently
            boxes = prediction.boxes.xyxy.data
            confs = prediction.boxes.conf.data
            
        # Render bounding boxes fast using cached predictions
        if len(boxes) > 0:
            for i in range(len(boxes)):
                confidence = float(confs[i])
                x1, y1, x2, y2 = map(int, boxes[i])
                
                # Draw main frame boundary box
                cv2.rectangle(frame, (x1, y1), (x2, y2), box_color, 2)
                
                # Draw text and label banner efficiently
                label = f"{class_name}: {confidence:.2f}"
                cv2.rectangle(frame, (x1, y1 - 18), (x1 + 95, y1), box_color, -1)
                cv2.putText(frame, label, (x1 + 2, y1 - 5), cv2.FONT_HERSHEY_SIMPLEX, 0.4, (255, 255, 255), 1, cv2.LINE_AA)

        # Calculate and overlay real-time FPS counter
        current_time = time.time()
        fps = 1 / (current_time - prev_time) if (current_time - prev_time) > 0 else 0
        prev_time = current_time
        cv2.putText(frame, f"FPS: {int(fps)}", (10, 30), cv2.FONT_HERSHEY_SIMPLEX, 0.7, (0, 255, 0), 2)

        #  Display frame
        cv2.imshow(window_name, frame)
        
        if cv2.waitKey(1) & 0xFF == ord('q'):
            print("🛑 Stop key 'q' pressed by user.")
            break

finally:
    cap.release()
    cv2.destroyAllWindows()
    cv2.waitKey(1)
    print("✨ Performance stream terminated clean.")
```
